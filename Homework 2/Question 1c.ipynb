```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def calculate_c(a, D = 1):
    return (np.sqrt(D) * (1 - 2*a)) / np.sqrt(a * (1 - a))
```

```python
x = np.linspace(0.01, 0.99, 100)
y = calculate_c(x)

plt.plot(x, y)

plt.xlabel("a")
plt.ylabel("c")

plt.axhline(y = 2, color = "red", linestyle = "--", label = "Passive Apparent Speed ")   
plt.axvline(x = (2 - np.sqrt(2)) / 4, color = "black", linestyle = "--", label = "Crossing (a $\\approx$ 0.1464)")

plt.legend()
plt.show()
```
